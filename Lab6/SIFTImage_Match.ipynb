```python
import cv2
import numpy as np

# ============================================================
# STEP 1: Load the Right Image
# ============================================================

img = cv2.imread('assets/TowerRight.jpg')

# Check if image was loaded successfully
if img is None:
    print("Error: Could not load the first image.")
    exit()

# Convert the image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)


# ============================================================
# STEP 2: Create SIFT Object
# ============================================================

sift = cv2.SIFT_create()


# ============================================================
# STEP 3: Detect Keypoints and Compute Descriptors
# ============================================================

keypoints, descriptors = sift.detectAndCompute(gray, None)


# ============================================================
# STEP 4: Draw SIFT Keypoints
# ============================================================

img_with_keypoints = cv2.drawKeypoints(
    img,
    keypoints,
    None,
    flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
)

# Display the image with keypoints
cv2.imshow('SIFT Keypoints', img_with_keypoints)
cv2.waitKey(0)
cv2.destroyAllWindows()


# ============================================================
# STEP 5: Print Keypoint and Descriptor Information
# ============================================================

print("Number of keypoints detected:", len(keypoints))

if descriptors is not None:
    print("Shape of descriptors:", descriptors.shape)
else:
    print("No descriptors were detected.")


# ============================================================
# STEP 6: Load the Second Image
# ============================================================

img2 = cv2.imread('assets/TowerLeft.jpg')

# Check if second image was loaded successfully
if img2 is None:
    print("Error: Could not load the second image.")
    exit()

# Convert second image to grayscale
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)


# ============================================================
# STEP 7: Detect SIFT Keypoints and Descriptors in Image 2
# ============================================================

keypoints2, descriptors2 = sift.detectAndCompute(gray2, None)

print("Number of keypoints detected in second image:", len(keypoints2))


# ============================================================
# STEP 8: Create Brute-Force Matcher
# ============================================================

bf = cv2.BFMatcher()

# Find the two best matches for each descriptor
matches = bf.knnMatch(descriptors, descriptors2, k=2)


# ============================================================
# STEP 9: Apply Lowe's Ratio Test
# ============================================================

good_matches = []

for m, n in matches:
    if m.distance < 0.75 * n.distance:
        good_matches.append([m])


# Print number of good matches
print("Number of good matches:", len(good_matches))


# ============================================================
# STEP 10: Draw Good Matches
# ============================================================

img_matches = cv2.drawMatchesKnn(
    img,
    keypoints,
    img2,
    keypoints2,
    good_matches,
    None,
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)

# Display matches
cv2.imshow("SIFT Good Matches", img_matches)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
